### an image with 3 channels is expected to have 3 dimensions ex. shape might look like [3,1024,768]
Axis 0 (Size 3): The color channels (typically Red, Green, Blue).

Axis 1 (Size 1024): The height of the image (number of rows).

Axis 2 (Size 768): The width of the image (number of columns).

### for the below case 
Axis 0 (Size 1): You still only have one main block of data wrapped in the outermost brackets.

Axis 1 (Size 4): Inside that block, you now have four rows instead of three.

Axis 2 (Size 2): Each row still contains exactly two numbers.

```python
T3 = torch.tensor([[[1, 2],
                        [3, 6],
                        [2, 4],[5,6]]])
```

```python
T3.ndim, T3.shape
```

```python
T4 = torch.tensor([[[1, 2],
                    [3, 6],
                    [2, 4],
                    [5,6]],
                  [[1, 2,3],
                    [3, 6,3],
                    [2, 4,3],
                    [5,6,3]]
                  ])
```

### the structure can be read as 
axis 0: has 2 elements
axis 1: each of the 2 elements' size which is 4
axis 2: each of the 4 elements' size which is 2

```python
# increasing the axis 0 by 1
T4 = torch.tensor([
                  [[1, 2],
                    [3, 6],
                    [2, 4],
                    [5,6]],
                  [[1, 2],
                    [3, 6],
                    [2, 4],
                    [5,6]]
                  ])
T4.ndim, T4.shape
```

# Reshaping, stacking, squeezing and unsqueezing

```python
neu = torch.arange(5,25,2)
neu.shape
```

```python
neu
```

```python
neu = neu.reshape(2,5)
neu
```

```python
neu.shape
```

```python
neu_stacked_0 = torch.stack([neu,neu,neu],dim=0)
neu_stacked_0, neu_stacked_0.ndim, neu_stacked_0.shape
```

```python
neu_stacked_1 = torch.stack([neu,neu,neu],dim=1)
neu_stacked_1, neu_stacked_1.ndim, neu_stacked_1.shape
```

```python
neu_stacked_2 = torch.stack([neu,neu,neu],dim=2)
neu_stacked_2, neu_stacked_2.ndim, neu_stacked_2.shape
```

```python
# Original tensor shape: (32, 224, 224, 3)
x = torch.randn(32, 224, 224, 3)

# High-dimensional transpose using permute
# Reorders: Dim 0->0, Dim 3->1, Dim 1->2, Dim 2->3
y = x.permute(0, 3, 1, 2) 

print(y.shape) # Output: torch.Size([32, 3, 224, 224])
```
